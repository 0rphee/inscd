```python

events = sb.events(match_id=3857255)
events.shape
```

```python
fig, axs = plt.subplots(figsize=(10,4))
sns.heatmap(events.isna(), ax=axs, cbar=False, cmap="Blues")
plt.show()
```

```python
for col in events.columns:
    if 'pass' in col:
        print(col, end=", ")
```

```python
for col in events.columns:
    if 'pass' not in col:
        print(col, end=", ")
```

```python
variables = ["minute", 'second', 'location', 'pass_end_location', 'player', 'pass_recipient', 'team', 'type']
```

```python
passes = events[variables]
passes.head(10)
```

```python
passes['type'].unique()
```

```python
passes[passes['type'] == "Pass"]
```

```python
final = passes[passes['type'] == 'Pass']
final.reset_index(drop=True, inplace=True)
final.shape
```

```python
final
```

```python
final.location.dtype
```

```python
final.location[0]
```

```python
final['x0'] = final.location.apply(lambda x: x[0])
final['y0'] = final.location.apply(lambda x: x[1])
final['x1'] = final.pass_end_location.apply(lambda x: x[0])
final['x1'] = final.pass_end_location.apply(lambda x: x[1])
final.drop(columns=["location", "pass_end_location"], inplace=True)
final.head()
```

```python
sns.scatterplot(data=final, x='x0', y='y0')
plt.show()
```

```python
from mplsoccer import Pitch

pitch = Pitch(pitch_color='grass', line_color='white', stripe=True)
fig, axs = pitch.draw()

sns.scatterplot(data=final, x='x0', y='y0', ax=axs)
plt.show()
```

```python
def plot_minute(minute):
    pitch = Pitch(pitch_color='grass', line_color='white', stripe=True)
    fig, axs = pitch.draw()

    sns.scatterplot(data=final[final.minute==minute], x='x0', y='y0', ax=axs, hue='team')
    plt.legend(loc='upper center', ncols=2)
    plt.show()
```

```python
plot_minute(0)
```

```python
import ipywidgets as widgets

widgets.interact(plot_minute, minute=(0, 90,1))
```
