```python
# https://www.nytimes.com/2026/03/26/learning/whats-going-on-in-this-graph-april-8-2026.html
from matplotlib.patches import Rectangle
# Data
elementary = {
    "≤1 hour": 68,
    "1-2 hours": 19,
    "2-3 hours": 10,
    "3+ hours": 3
}
middle_high = {
    "≤1 hour": 13,
    "1-2 hours": 22,
    "2-3 hours": 24,
    "3+ hours": 40
}
# Colors sampled from the original image
YELLOW = "#FBC011"
GRAY = "#CCCCCC"
DARK = "#121212"
WHITE = "#FFFFFF"
# Figure size matching the original image: 934 × 564 px
fig, ax = plt.subplots(figsize=(9.34, 5.64), dpi=100)
fig.patch.set_facecolor(WHITE)
ax.set_facecolor(WHITE)
# Coordinate system approximating the original pixel layout
ax.set_xlim(0, 934)
ax.set_ylim(564, 0)
ax.axis("off")
# -------------------------------------------------------------------
# Typography
# -------------------------------------------------------------------
FONT = "Arial"
# Main title
ax.text(
    16, 36,
    "Time students spend on devices",
    fontsize=23,
    fontweight="bold",
    color=DARK,
    fontfamily=FONT,
    va="center"
)
# Section titles
ax.text(
    16, 76,
    "Elementary school",
    fontsize=20,
    fontweight="bold",
    color="#333333",
    fontfamily=FONT,
    va="center"
)
ax.text(
    16, 330,
    "Middle and high school",
    fontsize=20,
    fontweight="bold",
    color="#333333",
    fontfamily=FONT,
    va="center"
)
# -------------------------------------------------------------------
# Bars
# -------------------------------------------------------------------
bar_x = 136
bar_height = 43
# Scaling from the original image:
# 68% ≈ 780 px, therefore ≈ 11.47 px per percentage point
scale = 11.47
# Original bar positions
elementary_y = [109, 155, 201, 246]
middle_y = [364, 410, 456, 501]
labels = list(elementary.keys())
def draw_section(data, ys, highlighted_index):
    for i, (label, value) in enumerate(data.items()):
        y = ys[i]
        # Label
        ax.text(
            16, y + bar_height / 2,
            label,
            fontsize=18,
            color="#121212",
            fontfamily=FONT,
            va="center"
        )
        # Bar
        width = value * scale
        color = YELLOW if i == highlighted_index else GRAY
        ax.add_patch(
            Rectangle(
                (bar_x, y),
                width,
                bar_height,
                facecolor=color,
                edgecolor="none"
            )
        )
        # Percentage
        # For the highlighted 68% bar, the percentage is inside the bar.
        if i == highlighted_index:
            ax.text(
                bar_x + width - 13,
                y + bar_height / 2,
                f"{value}%",
                fontsize=18,
                fontweight="bold",
                color="#000000",
                fontfamily=FONT,
                ha="right",
                va="center"
            )
        else:
            ax.text(
                bar_x + width + 8,
                y + bar_height / 2,
                f"{value}%",
                fontsize=18,
                color="#121212",
                fontfamily=FONT,
                ha="left",
                va="center"
            )
# Elementary: ≤1 hour highlighted
draw_section(elementary, elementary_y, highlighted_index=0)
# Middle/high: 3+ hours highlighted
draw_section(middle_high, middle_y, highlighted_index=3)
# -------------------------------------------------------------------
# Render
# -------------------------------------------------------------------
plt.subplots_adjust(left=0, right=1, top=1, bottom=0)
plt.show()
# To save:
# plt.savefig("replica.png", dpi=100, bbox_inches="tight", pad_inches=0)
```
