```python
stats =  {"Babysitters": 98, "Preschool teachers": 97, "Child care workers": 94, "Speech pathologists": 94, "School psychologists": 88, "Child social workers": 88, "Librarians": 83, "Solo parents": 82, "K-8 teachers": 79, "Guidance counselors": 79, "Tutors": 70, "Pediatricians": 69, "Volunteer mentors": 60, "High school teachers": 57, "School bus drivers": 56, "Grandparent caregivers": 54, "Scout leaders": 33, "Coaches": 26, "Religious leaders": 14}
stats2 = {key: 100-val for key, val in stats.items()}

colors = {"BDARK": "#3A64A1", "BLIGHT": "#BBDFFF"}

font_size = 9
TITLE_FONT = "Liberation Serif"
FONT = "Liberation Sans"
plt.rcParams['font.family'] = FONT

fig, ax = plt.subplots(figsize=(4,6))
bottom = [0 for _ in range(len(stats))]
width = 0.7
for c, s in zip(colors.values(), (stats, stats2)):
    p = ax.barh(s.keys(), s.values(), width, left=bottom, color=c)
    bottom = [original + adicional for original, adicional in zip(bottom, s.values())]

# Add percentage numbers inside the bars
for i, (key, val) in enumerate(stats.items()):
    ax.text(2, i, f"{val}%", color="white", va="center", ha="left", fontsize=font_size-1)
    val2 = stats2[key]
    if val2 > 10:
        ax.text(98, i, f"{val2}%", color="black", va="center", ha="right", fontsize=font_size-1)
ax.invert_yaxis()
ax.set_xticks([])

ax.tick_params(axis="y", left=False, pad=125, labelsize=font_size)
for label in ax.get_yticklabels():
    label.set_horizontalalignment("left")

for spine in ax.spines.values():
    spine.set_visible(False)
ax.set_title("The Adults in Children's Lives", ha="left", x=-0.58, fontsize=font_size+1, fontweight="bold", fontfamily=TITLE_FONT)
ax.text(-0.58, 0.98, "The share of women (dark) and men (light) in each role:", ha="left", transform=ax.transAxes, fontsize=font_size, fontfamily=TITLE_FONT)
plt.show()

```
